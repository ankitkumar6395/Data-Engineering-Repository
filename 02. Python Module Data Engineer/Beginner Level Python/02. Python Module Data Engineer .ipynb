## 📚 STUDY – Theory Day 3
### 1. What Are Operators?
Operators are symbols or keywords used to perform operations on values and variables.

````python
salary = 40000
bonus = 5000

total = salary + bonus

print(total)
````

### 2. Arithmetic Operators

Arithmetic operators are used for calculations.

| Operator | Meaning        | Example   |
|----------|----------------|-----------|
| `+`      | Addition       | `10 + 5`  |
| `-`      | Subtraction    | `10 - 5`  |
| `*`      | Multiplication | `10 * 5`  |
| `/`      | Division       | `10 / 5`  |
| `//`     | Floor Division | `11 // 2` |
| `%`      | Remainder      | `11 % 2`  |
| `**`     | Power          | `2 ** 3`  |

````python
a = 20
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

### 3. Comparison Operators

Comparison operators compare two values.

The result is either `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal |
| `!=` | Not Equal |
| `>` | Greater Than |
| `<` | Less Than |
| `>=` | Greater Than or Equal |
| `<=` | Less Than or Equal |

````python
salary = 50000

print(salary == 50000)
print(salary != 40000)
print(salary > 40000)
print(salary < 60000)
print(salary >= 50000)
print(salary <= 50000)
````

### 4. Logical Operators
Logical operators are used to combine conditions.
####    and
Both conditions must be True.
####     or
At least one condition must be True.
####     not
Changes True to False and False to True.

````python
age = 25
salary = 40000

print(age >= 18 and salary > 0)
print(age < 18 or salary > 30000)
print(not age < 18)
````

### 5. Assignment Operators

Assignment operators are used to assign or update values.

| Operator | Example |
|---|---|
| `=` | `salary = 30000` |
| `+=` | `salary += 5000` |
| `-=` | `salary -= 2000` |
| `*=` | `salary *= 2` |
| `/=` | `salary /= 2` |

````python
salary = 30000

salary += 5000
print(salary)

salary -= 2000
print(salary)
````

### 6. Membership Operators
Membership operators check whether a value exists in a collection.
Two important operators are:
- in
- not in

````python
departments = ["IT", "HR", "Finance"]

print("IT" in departments)
print("Sales" in departments)
print("Sales" not in departments)
````

#### 🔄 Operator Flow
    Data
    ↓  
    Choose Operator
    ↓  
    Perform Operation
    ↓  
    Get Result
    
##### For example:
Salary → Comparison Operator → salary > 50000 → True/False

2. Check whether age is valid.
3. Check whether salary is positive.

### 1: Create Employee Data

````python
# Employee information

employee_name = "Ankit"
employee_age = 25
department = "IT"
monthly_salary = 40000

allowed_departments = ["IT", "HR", "Finance"]

print("Employee Name:", employee_name)
print("Age:", employee_age)
print("Department:", department)
print("Monthly Salary:", monthly_salary)
````

````python
# Calculate annual salary

annual_salary = monthly_salary * 12

print("Annual Salary:", annual_salary)
````

## 📚 STUDY – Theory DAY 4
### 🔹 1. What Are Conditional Statements?
Conditional statements allow Python to execute different code depending on whether a condition is True or False.
For example, while processing employee data, we may want to check whether an employee's salary is valid.

````python
salary = 50000

if salary > 0:
    print("Valid Salary")
````

### 🔹 2. if Statement
The if statement executes a block of code when its condition is True.

````python
age = 25

if age >= 18:
    print("Employee is eligible")
````

If age >= 18 is True, Python executes the print() statement.
### 🔹 3. if + else
When there are two possible outcomes, we can use else.

````python
salary = 25000

if salary >= 30000:
    print("Salary is acceptable")
else:
    print("Salary is below the required amount")
````

### 🔹 4. if + elif + else
When we have multiple possible conditions, we can use elif.

````python
salary = 45000

if salary >= 60000:
    print("High Salary")
elif salary >= 30000:
    print("Medium Salary")
else:
    print("Low Salary")
````

### 🔹 5. Multiple Conditions
We can combine conditions using logical operators.

````python
age = 25
salary = 40000

if age >= 18 and salary > 0:
    print("Valid Employee")
else:
    print("Invalid Employee")
````

This is useful when a data record must satisfy multiple requirements.
### 🔹 6. Nested if
A nested if means using one if statement inside another if statement.

````python
age = 25
department = "IT"

if age >= 18:
    if department == "IT":
        print("Valid IT Employee")
````

### Topic 1: What Is a Loop?
A loop repeats a block of code multiple times. It reduces repetitive coding and helps process multiple records.

````python
employees = ["Ankit", "Rahul", "Priya"]

for employee in employees:
    print(employee)
````

Data Engineering use: Processing employee records, files, and data items.
### Topic 2: for Loop and range()
A for loop iterates over items in a sequence or other iterable.

````python
for number in range(1, 6):
    print(number)
````

The range(1, 6) function generates numbers from 1 to 5. The stop value is excluded.
```text
Expression	Meaning
range(5)	0 to 4
range(1, 6)	1 to 5
range(2, 11, 2)	2, 4, 6, 8, 10
```
### Topic 3: while Loop
A while loop runs as long as its condition is True.

````python
count = 1

while count <= 3:
    print(count)
    count += 1
````

The counter must be updated appropriately to avoid an infinite loop.
### Topic 4: break and continue
- break stops the loop.
- continue skips the current iteration and moves to the next one.

````python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
````

These statements can help control which records are processed.
### Topic 5: Counters, Accumulators and Nested Loops
A counter counts items. An accumulator maintains a running total.

````python
salaries = [20000, 30000, 40000]
total = 0

for salary in salaries:
    total += salary

print("Total Salary:", total)
````

- A nested loop is a loop inside another loop. It can process combinations of items from different groups.


````python
departments = ["IT", "HR"]

for department in departments:
    for shift in ["Morning", "Evening"]:
        print(department, shift)
````

🛠️ 5. MINI PROJECT — Real-World Data Pipeline Batch Processing System
##### 📋 Business Problem
A company receives employee records from an upstream system. Some records have invalid salary values. The data engineering 
#### team must:
- Process valid records.
- Reject records with salaries less than or equal to zero.
- Count processed and rejected records.
- Calculate total payroll and average valid salary.
- Generate a batch report.
💻 Code Cell 1: Employee Batch Data

````python
employees = [
    {"name": "Ankit", "department": "Data Engineering", "salary": 40000},
    {"name": "Rahul", "department": "IT", "salary": 35000},
    {"name": "Priya", "department": "HR", "salary": 0},
    {"name": "Aman", "department": "Finance", "salary": 50000},
    {"name": "Neha", "department": "IT", "salary": -5000}
]

processed_count = 0
rejected_count = 0
total_salary = 0

print("===== BATCH PROCESSING STARTED =====")

for employee in employees:
    salary = employee["salary"]

    if salary <= 0:
        rejected_count += 1
        print("REJECTED:", employee["name"], "- Invalid salary")
        continue

    processed_count += 1
    total_salary += salary

    print("PROCESSED:", employee["name"], "-", salary)

average_salary = (
    total_salary / processed_count
    if processed_count > 0
    else 0
)

print("\n===== BATCH PROCESSING REPORT =====")
print("Total input records:", len(employees))
print("Successfully processed:", processed_count)
print("Rejected records:", rejected_count)
print("Total valid payroll: ₹", total_salary)
print("Average valid salary: ₹", average_salary)
print("===== BATCH PROCESSING COMPLETED =====")
````

### 2: Process and Validate Records

````python
processed_count = 0
rejected_count = 0
total_salary = 0

print("===== BATCH PROCESSING STARTED =====")

for employee in employees:
    salary = employee["salary"]

    if salary <= 0:
        rejected_count += 1
        print("REJECTED:", employee["name"], "- Invalid salary")
        continue

    processed_count += 1
    total_salary += salary

    print("PROCESSED:", employee["name"], "-", salary)
````

### 3: Generate the Report

````python
average_salary = (
    total_salary / processed_count
    if processed_count > 0
    else 0
)

print("\n===== BATCH PROCESSING REPORT =====")
print("Input Records:", len(employees))
print("Processed Records:", processed_count)
print("Rejected Records:", rejected_count)
print("Valid Payroll: ₹", total_salary)
print("Average Valid Salary: ₹", round(average_salary, 2))
print("===== BATCH PROCESSING COMPLETED =====")
````
