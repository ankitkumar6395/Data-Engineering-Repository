## 📚 STUDY – Theory Day 3
### 1. What Are Operators?
Operators are symbols or keywords used to perform operations on values and variables.

````python
salary = 40000
bonus = 5000

total = salary + bonus

print(total)
````

### 2. Arithmetic Operators

Arithmetic operators are used for calculations.

| Operator | Meaning        | Example   |
|----------|----------------|-----------|
| `+`      | Addition       | `10 + 5`  |
| `-`      | Subtraction    | `10 - 5`  |
| `*`      | Multiplication | `10 * 5`  |
| `/`      | Division       | `10 / 5`  |
| `//`     | Floor Division | `11 // 2` |
| `%`      | Remainder      | `11 % 2`  |
| `**`     | Power          | `2 ** 3`  |

````python
a = 20
b = 5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

### 3. Comparison Operators

Comparison operators compare two values.

The result is either `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal |
| `!=` | Not Equal |
| `>` | Greater Than |
| `<` | Less Than |
| `>=` | Greater Than or Equal |
| `<=` | Less Than or Equal |

````python
salary = 50000

print(salary == 50000)
print(salary != 40000)
print(salary > 40000)
print(salary < 60000)
print(salary >= 50000)
print(salary <= 50000)
````

### 4. Logical Operators
Logical operators are used to combine conditions.
####    and
Both conditions must be True.
####     or
At least one condition must be True.
####     not
Changes True to False and False to True.

````python
age = 25
salary = 40000

print(age >= 18 and salary > 0)
print(age < 18 or salary > 30000)
print(not age < 18)
````

### 5. Assignment Operators

Assignment operators are used to assign or update values.

| Operator | Example |
|---|---|
| `=` | `salary = 30000` |
| `+=` | `salary += 5000` |
| `-=` | `salary -= 2000` |
| `*=` | `salary *= 2` |
| `/=` | `salary /= 2` |

````python
salary = 30000

salary += 5000
print(salary)

salary -= 2000
print(salary)
````

### 6. Membership Operators
Membership operators check whether a value exists in a collection.
Two important operators are:
- in
- not in

````python
departments = ["IT", "HR", "Finance"]

print("IT" in departments)
print("Sales" in departments)
print("Sales" not in departments)
````

#### 🔄 Operator Flow
    Data
    ↓  
    Choose Operator
    ↓  
    Perform Operation
    ↓  
    Get Result
    
##### For example:
Salary → Comparison Operator → salary > 50000 → True/False

### 🔹 1. What Are Conditional Statements?
Conditional statements allow Python to execute different code depending on whether a condition is True or False.
For example, while processing employee data, we may want to check whether an employee's salary is valid.

````python
salary = 50000

if salary > 0:
    print("Valid Salary")
````

### 🔹 2. if Statement
The if statement executes a block of code when its condition is True.

````python
age = 25

if age >= 18:
    print("Employee is eligible")
````

If age >= 18 is True, Python executes the print() statement.
### 🔹 3. if + else
When there are two possible outcomes, we can use else.

````python
salary = 25000

if salary >= 30000:
    print("Salary is acceptable")
else:
    print("Salary is below the required amount")
````

### 🔹 4. if + elif + else
When we have multiple possible conditions, we can use elif.

````python
salary = 45000

if salary >= 60000:
    print("High Salary")
elif salary >= 30000:
    print("Medium Salary")
else:
    print("Low Salary")
````

### 🔹 5. Multiple Conditions
We can combine conditions using logical operators.

````python
age = 25
salary = 40000

if age >= 18 and salary > 0:
    print("Valid Employee")
else:
    print("Invalid Employee")
````

This is useful when a data record must satisfy multiple requirements.
### 🔹 6. Nested if
A nested if means using one if statement inside another if statement.

````python
age = 25
department = "IT"

if age >= 18:
    if department == "IT":
        print("Valid IT Employee")
````

🛠️ 5. MINI PROJECT — Real-World Data Pipeline Batch Processing System
##### 📋 Business Problem
A company receives employee records from an upstream system. Some records have invalid salary values. The data engineering 
#### team must:
- Process valid records.
- Reject records with salaries less than or equal to zero.
- Count processed and rejected records.
- Calculate total payroll and average valid salary.
- Generate a batch report.
💻 Code Cell 1: Employee Batch Data

````python
employees = [
    {"name": "Ankit", "department": "Data Engineering", "salary": 40000},
    {"name": "Rahul", "department": "IT", "salary": 35000},
    {"name": "Priya", "department": "HR", "salary": 0},
    {"name": "Aman", "department": "Finance", "salary": 50000},
    {"name": "Neha", "department": "IT", "salary": -5000}
]

processed_count = 0
rejected_count = 0
total_salary = 0

print("===== BATCH PROCESSING STARTED =====")

for employee in employees:
    salary = employee["salary"]

    if salary <= 0:
        rejected_count += 1
        print("REJECTED:", employee["name"], "- Invalid salary")
        continue

    processed_count += 1
    total_salary += salary

    print("PROCESSED:", employee["name"], "-", salary)

average_salary = (
    total_salary / processed_count
    if processed_count > 0
    else 0
)

print("\n===== BATCH PROCESSING REPORT =====")
print("Total input records:", len(employees))
print("Successfully processed:", processed_count)
print("Rejected records:", rejected_count)
print("Total valid payroll: ₹", total_salary)
print("Average valid salary: ₹", average_salary)
print("===== BATCH PROCESSING COMPLETED =====")
````

### 2: Process and Validate Records

````python
processed_count = 0
rejected_count = 0
total_salary = 0

print("===== BATCH PROCESSING STARTED =====")

for employee in employees:
    salary = employee["salary"]

    if salary <= 0:
        rejected_count += 1
        print("REJECTED:", employee["name"], "- Invalid salary")
        continue

    processed_count += 1
    total_salary += salary

    print("PROCESSED:", employee["name"], "-", salary)
````

### 3: Generate the Report

````python
average_salary = (
    total_salary / processed_count
    if processed_count > 0
    else 0
)

print("\n===== BATCH PROCESSING REPORT =====")
print("Input Records:", len(employees))
print("Processed Records:", processed_count)
print("Rejected Records:", rejected_count)
print("Valid Payroll: ₹", total_salary)
print("Average Valid Salary: ₹", round(average_salary, 2))
print("===== BATCH PROCESSING COMPLETED =====")
````

## 🛠️ 5. MINI PROJECT — Employee Text Data Cleaning System
- 📋 Real-World Problem
A company receives employee data from different sources. Some records contain extra spaces, inconsistent capitalisation, or invalid email formats.
- The data engineering team needs to standardise the text fields and identify records that need attention.
###### Project requirements:
- Remove extra spaces from employee names and departments.
- Standardise capitalisation.
- Convert email addresses to lowercase.
- Check for basic email-format issues.
- Generate a cleaned employee report.
- Count records that need review.
### 1 — Input Data

````python
employees = [
    {
        "name": "  ankit kumar ",
        "department": " data engineering ",
        "email": "ANKIT@EXAMPLE.COM"
    },
    {
        "name": "rahul sharma",
        "department": " IT ",
        "email": "rahul.example.com"
    },
    {
        "name": " PRIYA SINGH ",
        "department": "finance",
        "email": " priya@example.com "
    },
    {
        "name": "aman verma",
        "department": " DATA ENGINEERING ",
        "email": "aman@company.org"
    }
]
````

### 2 — Clean and Validate Text

````python
cleaned_employees = []
review_count = 0

for employee in employees:
    name = employee["name"].strip().title()
    department = employee["department"].strip().title()
    email = employee["email"].strip().lower()

    email_valid = (
        "@" in email
        and email.count("@") == 1
        and "." in email.split("@")[-1]
        and " " not in email
    )

    if not name or not department or not email_valid:
        status = "Needs Review"
        review_count += 1
    else:
        status = "Clean"

    cleaned_employees.append({
        "name": name,
        "department": department,
        "email": email,
        "status": status
    })
````

### 3 — Generate the Cleaning Report

````python
print("=" * 60)
print("       EMPLOYEE TEXT DATA CLEANING REPORT")
print("=" * 60)

for employee in cleaned_employees:
    print("Name       :", employee["name"])
    print("Department :", employee["department"])
    print("Email      :", employee["email"])
    print("Status     :", employee["status"])
    print("-" * 60)

print("Total Records :", len(employees))
print("Needs Review  :", review_count)
print("Clean Records :", len(employees) - review_count)
print("=" * 60)
````

A company needs a simple system to manage employee records without manually changing the source code every time.
- The program should allow users to view, add, search, update, and remove employee records, and count the total employees.
- 💡 Hint: Use a list of dictionaries to store employees, a while loop for the menu, and if-elif-else conditions to perform each operation.


````python
# Employee Records Management System

employees = [
    {"id": 101, "name": "Amit", "department": "IT", "salary": 45000},
    {"id": 102, "name": "Priya", "department": "Finance", "salary": 55000},
    {"id": 103, "name": "Rahul", "department": "HR", "salary": 40000}
]

while True:
    print("\n===== Employee Records Management System =====")
    print("1. View All Employees")
    print("2. Add Employee")
    print("3. Search Employee")
    print("4. Update Employee Salary")
    print("5. Remove Employee")
    print("6. Count Employees")
    print("7. Exit")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        if employees:
            for employee in employees:
                print(employee)
        else:
            print("No employee records found.")

    elif choice == "2":
        try:
            emp_id = int(input("Enter employee ID: "))

            if any(emp["id"] == emp_id for emp in employees):
                print("Employee ID already exists.")
                continue

            name = input("Enter employee name: ").strip()
            department = input("Enter department: ").strip()
            salary = float(input("Enter salary: "))

            if not name or not department or salary <= 0:
                print("Invalid details. Name, department, and positive salary are required.")
                continue

            employees.append({
                "id": emp_id,
                "name": name,
                "department": department,
                "salary": salary
            })

            print("Employee added successfully.")

        except ValueError:
            print("Please enter a valid ID and salary.")

    elif choice == "3":
        try:
            emp_id = int(input("Enter employee ID to search: "))
            found = False

            for employee in employees:
                if employee["id"] == emp_id:
                    print("Employee found:", employee)
                    found = True
                    break

            if not found:
                print("Employee not found.")

        except ValueError:
            print("Please enter a valid employee ID.")

    elif choice == "4":
        try:
            emp_id = int(input("Enter employee ID: "))
            found = False

            for employee in employees:
                if employee["id"] == emp_id:
                    new_salary = float(input("Enter new salary: "))

                    if new_salary > 0:
                        employee["salary"] = new_salary
                        print("Salary updated successfully.")
                    else:
                        print("Salary must be greater than zero.")

                    found = True
                    break

            if not found:
                print("Employee not found.")

        except ValueError:
            print("Please enter a valid ID and salary.")

    elif choice == "5":
        try:
            emp_id = int(input("Enter employee ID to remove: "))

            for employee in employees:
                if employee["id"] == emp_id:
                    employees.remove(employee)
                    print("Employee removed successfully.")
                    break
            else:
                print("Employee not found.")

        except ValueError:
            print("Please enter a valid employee ID.")

    elif choice == "6":
        print("Total employees:", len(employees))

    elif choice == "7":
        print("Exiting Employee Records Management System.")
        break

    else:
        print("Invalid choice. Please select 1–7.")
````

````python
employees = [
    {"department": "IT", "salary": 50000},
    {"department": "HR", "salary": 40000},
    {"department": "IT", "salary": 60000},
    {"department": "HR", "salary": 50000}
]

salary_totals = {}
employee_counts = {}

for employee in employees:
    department = employee["department"]

    salary_totals[department] = (
        salary_totals.get(department, 0) + employee["salary"]
    )

    employee_counts[department] = (
        employee_counts.get(department, 0) + 1
    )

average_salary = {}

for department in salary_totals:
    average_salary[department] = (
        salary_totals[department] / employee_counts[department]
    )

print(average_salary)
````

### 🛠️ 5. MINI PROJECT — Employee Data Quality and Summary System
##### 📋 Real-World Problem
An ETL pipeline receives employee records from a source system. Some records may have missing fields, invalid salaries, or duplicate employee IDs. Before loading the data into a target database, the pipeline needs to validate the records and generate a summary report.

Project requirements:
- Store employee records as dictionaries inside a list.
- Identify duplicate employee IDs.
- Detect missing names and departments.
- Identify invalid salaries.
- Separate valid records from invalid records.
- Generate a summary of the results.
💡 Hint: Use a list of dictionaries, a set for duplicate detection, conditional statements for validation, and counters for the report.


````python
# Employee Data Quality and Summary System

employees = [
    {
        "id": 101,
        "name": "Ankit",
        "department": "IT",
        "salary": 45000
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "Finance",
        "salary": 55000
    },
    {
        "id": 103,
        "name": "",
        "department": "HR",
        "salary": 40000
    },
    {
        "id": 102,
        "name": "Priya Duplicate",
        "department": "Finance",
        "salary": 55000
    },
    {
        "id": 104,
        "name": "Rahul",
        "department": "",
        "salary": -2000
    },
    {
        "id": 105,
        "name": "Neha",
        "department": "IT",
        "salary": 60000
    }
]

seen_ids = set()
valid_records = []
invalid_records = []
duplicate_records = []

for employee in employees:
    employee_id = employee.get("id")
    issues = []

    # Check duplicate employee IDs
    if employee_id in seen_ids:
        issues.append("Duplicate employee ID")
        duplicate_records.append(employee)
    else:
        seen_ids.add(employee_id)

    # Check required fields
    if not employee.get("name", "").strip():
        issues.append("Missing name")

    if not employee.get("department", "").strip():
        issues.append("Missing department")

    # Check salary
    salary = employee.get("salary")

    if not isinstance(salary, (int, float)) or isinstance(salary, bool):
        issues.append("Invalid salary")
    elif salary <= 0:
        issues.append("Invalid salary")

    # Classify the record
    if issues:
        invalid_records.append({
            "record": employee,
            "issues": issues
        })
    else:
        valid_records.append(employee)

# Generate report
print("=" * 45)
print("EMPLOYEE DATA QUALITY REPORT")
print("=" * 45)

print("Total records:", len(employees))
print("Valid records:", len(valid_records))
print("Invalid records:", len(invalid_records))
print("Duplicate ID records:", len(duplicate_records))

print("\n--- Valid Employee Records ---")

for employee in valid_records:
    print(employee)

print("\n--- Invalid Employee Records ---")

for item in invalid_records:
    print("Record:", item["record"])
    print("Issues:", ", ".join(item["issues"]))

print("\n--- Duplicate Employee Records ---")

for employee in duplicate_records:
    print(employee)

print("\nData quality report generated successfully.")
````
