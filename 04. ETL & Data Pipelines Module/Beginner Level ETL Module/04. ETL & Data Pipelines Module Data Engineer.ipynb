## 📚 Study / Theory
### 1. What is ETL?
ETL stands for:
- E — Extract
- T — Transform
- L — Load
ETL is a process used to collect data from different sources, transform the data into a useful and consistent format, and load the processed data into a destination system.
- Source → Extract → Transform → Load → Destination

### 2. What is Extraction?
Extraction means collecting data from a source system.
Common data sources include:
- CSV files
- JSON files
- Excel files
- MySQL databases
- APIs
- Applications
### Example:
      employees.csv
           ↓
        Extract
           ↓
      Employee Data

### 3. What is Transformation?
Transformation means changing raw data into a clean and useful format.
Common transformations include:
- Removing duplicates
- Handling missing values
- Changing data types
- Renaming columns
- Filtering records
- Standardizing values
- Calculating new columns
### Example:    
    Raw Data
       ↓
    Remove duplicates
       ↓
    Handle missing values
       ↓
    Convert data types
       ↓
    Clean Data

### 4. What is Loading?
Loading means storing transformed data into a destination.
Common destinations include:
- MySQL
- PostgreSQL
- Data Warehouse
- Data Lake
- Cloud storage
### Example:
    Clean Data
        ↓
       LOAD
        ↓
    MySQL Database

ETL helps create clean, consistent, and usable data.

### 1.What Is a Data Pipeline?

A data pipeline is a series of steps used to move data from one system
to another.

During this process, data may be extracted, cleaned, transformed,
validated, and loaded into a destination.

## Example

A company receives employee information in a CSV file and needs to
Store the cleaned information in a MySQL database.

The pipeline reads the CSV file, processes the records, checks the
data, and loads the final records into MySQL.

### 2. Why Are Data Pipelines Important?

- Automate repetitive data tasks.
- Reduce manual work.
- Improve data consistency.
- Process large amounts of data.
- Make data available for reporting and analysis.

## 3. Types of Data Pipelines

### 1. Batch Pipeline

A batch pipeline processes a group of records together at scheduled
times or when a batch is available.

Example:
Processing all sales records at midnight.

### 2. Streaming Pipeline

A streaming pipeline processes data continuously or in small events
as new data arrives.

Example:
Processing payment events as customers make purchases.

### 3. Real-Time Pipeline

A real-time pipeline aims to process data with very low delay so
that the results are available quickly.

Example:
Detecting suspicious transactions shortly after they occur.
## 4. Batch Processing

Batch processing means collecting multiple records and processing
them together.

##### Example

A company receives 1,000 sales records every day.

The pipeline can:
1. Read the day's sales file.
2. Remove duplicate records.
3. Validate prices and quantities.
4. Calculate total sales.
5. Load the results into a database.

### Advantages

- Suitable for scheduled processing.
- Efficient for large groups of records.
- Easy to organize into daily or hourly jobs.

### Limitation

New data may not be available in the destination until the next
scheduled batch finishes.

## 4. 💻 Code Cells — Practical Examples Day 2
### 1 — Create a Data Pipeline

````python
# Source data
source_data = [10, 20, 30, 40, 50]

# Extract
extracted_data = source_data

# Transform
transformed_data = [value * 2 for value in extracted_data]

# Validate
validated_data = [
    value for value in transformed_data if value > 0
]

# Load
loaded_data = validated_data

print("Source Data:", source_data)
print("Transformed Data:", transformed_data)
print("Validated Data:", validated_data)
print("Loaded Data:", loaded_data)
````

### 2 — Validate Employee Records

````python
# Sample employee records
employees = [
    {"id": 1, "name": "Amit", "salary": 50000},
    {"id": 2, "name": "Priya", "salary": 60000},
    {"id": 3, "name": "", "salary": 45000},
    {"id": 4, "name": "Rahul", "salary": -1000}
]

# Extract
extracted_data = employees

# Transform and validate
valid_records = []

for employee in extracted_data:
    if employee["name"].strip() and employee["salary"] > 0:
        valid_records.append(employee)

# Load
loaded_data = valid_records

print("Total Records:", len(extracted_data))
print("Valid Records:", len(loaded_data))
print("Loaded Data:", loaded_data)
````

### 3 — Simulate Batch Processing

````python
# Daily sales records
sales = [100, 200, 150, 300, 250]

# Process the complete batch
total_sales = sum(sales)
number_of_records = len(sales)
average_sale = total_sales / number_of_records

print("Number of Records:", number_of_records)
print("Total Sales:", total_sales)
print("Average Sale:", average_sale)
````

## 🛠️ Mini Project — Daily Sales Processing Pipeline (Solution) Day 2
Objective: Build a Python ETL pipeline that validates sales records, calculates each valid sale, and generates a final sales report.
### 1: Create Sales Records

````python
# Source data: Daily sales records
sales_records = [
    {"product": "Laptop", "quantity": 2, "price": 50000},
    {"product": "Mouse", "quantity": 5, "price": 500},
    {"product": "", "quantity": 3, "price": 200},
    {"product": "Keyboard", "quantity": 0, "price": 1000},
    {"product": "Monitor", "quantity": 2, "price": 12000},
    {"product": "Headphones", "quantity": 3, "price": 1500},
    {"product": "Webcam", "quantity": 2, "price": -800}
]

print("Total Source Records:", len(sales_records))
````

### 2 — Extract and Validate Data

# Extract
extracted_data = sales_records

# Create lists for valid and rejected records
valid_sales = []
rejected_sales = []

# Validate each record
for record in extracted_data:
    if (
        record["product"].strip()
        and record["quantity"] > 0
        and record["price"] > 0
    ):
        valid_sales.append(record.copy())
    else:
        rejected_sales.append(record.copy())

print("Valid Records:", len(valid_sales))
print("Rejected Records:", len(rejected_sales))

### 3 — Transform the Data

````python
# Calculate the total amount for each valid sale
for record in valid_sales:
    record["total_amount"] = record["quantity"] * record["price"]

print("Transformed Sales Data:")

for record in valid_sales:
    print(record)
````

### 4 — Load and Generate the Report

````python
# Load the transformed data into the final in-memory dataset
loaded_data = valid_sales

# Calculate overall sales
total_sales = sum(
    record["total_amount"] for record in loaded_data
)

print("========== DAILY SALES REPORT ==========")

for record in loaded_data:
    print(
        f"Product: {record['product']} | "
        f"Quantity: {record['quantity']} | "
        f"Unit Price: ₹{record['price']} | "
        f"Total: ₹{record['total_amount']}"
    )

print("----------------------------------------")
print("Total Source Records:", len(extracted_data))
print("Valid Records:", len(loaded_data))
print("Rejected Records:", len(rejected_sales))
print(f"Overall Sales: ₹{total_sales}")
````
