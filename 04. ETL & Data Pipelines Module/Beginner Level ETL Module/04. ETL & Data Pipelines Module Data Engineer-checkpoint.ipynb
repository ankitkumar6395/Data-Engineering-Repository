## 📚 Study / Theory
### 1. What is ETL?
ETL stands for:
- E — Extract
- T — Transform
- L — Load
ETL is a process used to collect data from different sources, transform the data into a useful and consistent format, and load the processed data into a destination system.
- Source → Extract → Transform → Load → Destination

### 2. What is Extraction?
Extraction means collecting data from a source system.
Common data sources include:
- CSV files
- JSON files
- Excel files
- MySQL databases
- APIs
- Applications
### Example:
      employees.csv
           ↓
        Extract
           ↓
      Employee Data

### 3. What is Transformation?
Transformation means changing raw data into a clean and useful format.
Common transformations include:
- Removing duplicates
- Handling missing values
- Changing data types
- Renaming columns
- Filtering records
- Standardizing values
- Calculating new columns
### Example:    
    Raw Data
       ↓
    Remove duplicates
       ↓
    Handle missing values
       ↓
    Convert data types
       ↓
    Clean Data

### 4. What is Loading?
Loading means storing transformed data into a destination.
Common destinations include:
- MySQL
- PostgreSQL
- Data Warehouse
- Data Lake
- Cloud storage
### Example:
    Clean Data
        ↓
       LOAD
        ↓
    MySQL Database

### 5. Complete ETL Flow

### 6. Why is ETL Important?
ETL is important because real-world data is often:
- Incomplete
- Unstructured
- Duplicated
- Inconsistent
- Stored in different systems
ETL helps create clean, consistent, and usable data.

## 💻 Code Cells Day 1
### 1 — Simple ETL Representation
Type this manually in a Jupyter Code cell:

```python
# Define a simple source dataset
data = [10, 20, 30, 40, 50]

# Extract
extracted_data = data

# Transform
transformed_data = [value * 2 for value in extracted_data]

# Load
loaded_data = transformed_data

print("Extracted Data:", extracted_data)
print("Transformed Data:", transformed_data)
print("Loaded Data:", loaded_data)
```

### 2 — ETL with Employee Data

```python
# Source data
employees = [
    {"name": "Amit", "salary": 50000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Priya", "salary": 70000}
]

# Extract
extracted_data = employees

# Transform
for employee in extracted_data:
    employee["salary"] = employee["salary"] + 5000

# Load
loaded_data = extracted_data

print(loaded_data)
```
