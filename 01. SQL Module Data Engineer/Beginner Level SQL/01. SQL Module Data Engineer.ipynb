### Topic 1: What Is SQL?
SQL stands for Structured Query Language. It is used to communicate with relational databases.
Data Engineers use SQL to retrieve, insert, update, delete, and transform data.
```test

FROM employees;

### Topic 2: What Is a Database?
A database is an organized collection of data.
- For example, a company database may contain employee, customer, product, and order information.
```text
company_db
├── employees
├── customers
├── products
└── orders

### Topic 3: Tables, Rows, and Columns
A table stores related data in rows and columns.
employee_id	employee_name	department	salary
```text
101	Ankit	IT	50000
102	Rahul	HR	42000
103	Priya	Finance	55000
```
- Table: A collection of related records.
- Row: One complete record.
- Column: One attribute, such as department.
- Value: An individual piece of data, such as 50000.

### Topic 4: SQL Command Categories
Category	Purpose	Examples
DDL	defines database structures:	CREATE, ALTER, DROP.
DML	modifies data	INSERT, UPDATE, DELETE.
DQL	Retrieves data	SELECT
DCL	Controls permissions	GRANT, REVOKE
TCL	manages transactions	COMMIT, ROLLBACK.

### Topic 5: Why SQL Is Important for Data Engineers
SQL helps Data Engineers work with data stored in databases and data warehouses.
Common uses include:
- Extracting data from source tables
- Transforming data into useful formats
- Checking data quality
- Loading data into target tables
- Preparing datasets for analytics
