## Study: Theory Day 1

### Topic 1: What is Big Data?
Big Data refers to datasets that are very large, rapidly generated, diverse, or complex and cannot be efficiently handled using traditional data-processing methods alone.
Big Data technologies are designed to help organizations:
- Store huge amounts of data
- Process data efficiently
- Process data in parallel
- Handle continuously generated data
- Analyze data and extract useful information

#### Simple Example
An e-commerce company may generate:
- Customer information
- Product information
- Orders
- Website clicks
- Search history
- Payment transactions
- Customer reviews
- Application logs
When millions of customers generate this information continuously, the amount and complexity of data become too large for a simple single-machine system.

### Topic 2: Why Do We Need Big Data Technologies?
Traditional systems can work very well for normal-sized datasets.
However, problems arise when:
- Data becomes extremely large.
- Data arrives continuously.
- Data comes from many different sources.
- Data needs to be processed quickly.
- One computer cannot handle the workload.
Big Data technologies solve these problems using distributed storage and distributed processing.

### Traditional Processing
    Data
      ↓
    Single Computer
      ↓
    Processing
      ↓
    Result


### Distributed Processing
                     ┌── Computer 1 ──┐
    Data ────────────┼── Computer 2 ──┼── Processing
                     ├── Computer 3 ──┤
                     └── Computer 4 ──┘
                             ↓
                           Result

Instead of depending on one machine, the workload can be distributed across multiple machines.

### Topic 4: Types of Big Data
Big Data can broadly be divided into three categories.
##### Structured Data
Data with a predefined structure.
##### Example:
Employee_ID	Name	Department	Salary
101	Ankit	IT	60000
102	Rahul	HR	50000


Common technologies:
- MySQL
- PostgreSQL
- SQL Server
Semi-Structured Data
Data that does not follow a traditional table structure but contains some organization.
##### Examples:
- JSON
- XML
- Application logs
##### Example JSON:
    {
        "employee_id": 101,
        "name": "Ankit",
        "department": "IT"
    }

##### Unstructured Data
Data without a predefined tabular structure.
##### Examples:
- Images
- Videos
- Audio
- PDFs
- Emails
- Social media content

### Topic 5: Batch Processing vs Real-Time Processing
##### Batch Processing
Data is collected first and processed later.
    
    Data
     ↓
    Collect
     ↓
    Store
     ↓
    Process
     ↓
    Result

###### Example:
A company processes the previous day's sales every morning.
Common technologies:
- Hadoop
- Spark
- Hive
#### Real-Time Processing
Data is processed continuously as it arrives.

    Event
     ↓
    Streaming System
     ↓
    Processing
     ↓
    Immediate Result

###### Example:
A banking system detects suspicious transactions while transactions are happening.
Common technologies:
- Kafka
- Spark Structured Streaming

## Code Study Day 1
### 1: Import Pandas

```python
import pandas as pd
```

### 2: Create a Simple Dataset

```python
data = {
    "department": ["IT", "HR", "Finance", "IT", "Sales"],
    "employees": [120, 40, 60, 150, 80]
}

df = pd.DataFrame(data)

df
```

### 3: Inspect the Dataset

```python
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

df.info()
```

### 4: Calculate Total Employees

```python
total_employees = df["employees"].sum()

print("Total employees:", total_employees)
```

### 5: Find the Largest Department

```python
largest_department = df.loc[df["employees"].idxmax()]

print("Department:", largest_department["department"])
print("Employees:", largest_department["employees"])
```

### 7: Basic Visualization

```python
import matplotlib.pyplot as plt

df.plot(
    x="department",
    y="employees",
    kind="bar",
    title="Employees by Department"
)

plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

### 3. What does Volume mean?
Volume refers to the amount or size of data being generated and stored.

### 4. What is the difference between Velocity and Volume?
- Volume = How much data exists.
- Velocity = How quickly data is generated or processed.
###### Example:
If a website stores 10 TB of data, that represents Volume.
If the website receives 100,000 events every second, that represents Velocity.

### 5. What is Variety in Big Data?
Variety refers to the different types and formats of data.
###### Examples:
- Structured → SQL tables
- Semi-structured → JSON, XML
- Unstructured → Images, videos, audio, documents

### 6. What is Veracity?
Veracity refers to the accuracy, quality, consistency, and reliability of data.

### 7. What is the difference between structured and unstructured data?
Structured Data,	Unstructured Data

### 8. What is the difference between batch processing and real-time processing?
Batch processing processes data in groups after data has been collected.
###### Example: Processing yesterday's sales data every morning.
Real-time processing processes data continuously or with very low delay as it arrives.
###### Example: Detecting a suspicious banking transaction immediately after it occurs.

### 10. Why is distributed processing important for Big Data?
Distributed processing divides a large workload across multiple computers or processing nodes.
This helps:
- Process large datasets faster
- Handle workloads that one machine cannot handle efficiently
- Scale processing by adding more machines
- Improve fault tolerance

## 🛠️ MINI PROJECT: Employee Workforce Big Data Analysis 
We will solve it using Python + Pandas in separate Jupyter Notebook code cells.
### 1: Import Pandas

```python
import pandas as pd
```

### 2: Create Employee Dataset

```python
data = {
    "employee_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "employee_name": [
        "Ankit", "Rahul", "Priya", "Neha", "Amit",
        "Riya", "Vikas", "Pooja", "Rohit", "Sneha"
    ],
    "department": [
        "IT", "HR", "IT", "Finance", "IT",
        "HR", "Sales", "Finance", "IT", "Sales"
    ],
    "city": [
        "Delhi", "Noida", "Delhi", "Meerut", "Noida",
        "Delhi", "Delhi", "Noida", "Meerut", "Delhi"
    ],
    "salary": [
        65000, 50000, 72000, 60000, 80000,
        55000, 45000, 70000, 75000, 52000
    ]
}

df = pd.DataFrame(data)

df
```

### 3: Inspect the Data

```python
print("First 5 records:")
print(df.head())

print("\nDataset information:")
df.info()
```

### 4: Total Number of Employees

```python
total_employees = df["employee_id"].count()

print("Total Employees:", total_employees)
```

### 5: Total Salary

```python
total_salary = df["salary"].sum()

print("Total Salary:", total_salary)
```

### 6: Average Salary

```python
average_salary = df["salary"].mean()

print("Average Salary:", average_salary)
```

### 7: Highest Salary

```python
highest_salary = df["salary"].max()

print("Highest Salary:", highest_salary)
```

### 8: Lowest Salary

```python
lowest_salary = df["salary"].min()

print("Lowest Salary:", lowest_salary)
```

### 9: Department-wise Employee Count

```python
department_count = df.groupby("department")["employee_id"].count()

print(department_count)
```

### 10: Department-wise Average Salary

```python
department_average_salary = df.groupby("department")["salary"].mean()

print(department_average_salary)
```

### 11: City-wise Employee Count

```python
city_count = df.groupby("city")["employee_id"].count()

print(city_count)
```

## Final Business Analysis

```python
print("===== EMPLOYEE WORKFORCE ANALYSIS =====")

print("Total Employees:", df["employee_id"].count())
print("Total Salary:", df["salary"].sum())
print("Average Salary:", df["salary"].mean())
print("Highest Salary:", df["salary"].max())
print("Lowest Salary:", df["salary"].min())

print("\nDepartment-wise Employee Count:")
print(df.groupby("department")["employee_id"].count())

print("\nDepartment-wise Average Salary:")
print(df.groupby("department")["salary"].mean())

print("\nCity-wise Employee Count:")
print(df.groupby("city")["employee_id"].count())
```
