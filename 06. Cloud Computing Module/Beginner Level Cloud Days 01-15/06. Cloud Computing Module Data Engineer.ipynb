## Day 1: Introduction to Cloud Computing

### What is Cloud Computing?

Cloud computing means using computing resources over the internet instead of
maintaining all hardware and software on our own computer.

###### Examples of cloud resources:

- Servers
- Storage
- Databases
- Networking
- Applications
- Computing power
