## 📚 Study: Theory
### What is Workflow?
A workflow is a sequence of tasks executed in a particular order.
Extract → Transform → Load

### What is Workflow Orchestration?
Workflow orchestration means automatically managing tasks, their order, dependencies, scheduling, failures, and execution.
## Example:
```text
Extract
   ↓
Validate
   ↓
Transform
   ↓
Load
```
### What is a Task?
A task is one individual unit of work.
```text
Task 1 → Extract
Task 2 → Transform
Task 3 → Load
```
- Automation vs Orchestration
- Automation: Automatically runs a task.
- Orchestration: Coordinates multiple tasks.

### 6: Workflow with Data

````python
employees = [
    {"name": "Amit", "salary": 50000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Priya", "salary": 70000}
]


def extract_data():
    return employees


def transform_data(data):
    for employee in data:
        employee["salary"] = employee["salary"] * 1.10

    return data


def load_data(data):
    for employee in data:
        print(employee)


data = extract_data()
data = transform_data(data)
load_data(data)
````

### 7: Add Workflow Status

````python
def extract_data():
    print("Extract: SUCCESS")
    return True


def transform_data():
    print("Transform: SUCCESS")
    return True


def load_data():
    print("Load: SUCCESS")
    return True


def run_workflow():
    if extract_data():
        if transform_data():
            if load_data():
                print("Workflow completed successfully")


run_workflow()
````

### 8: Handle Failure

````python
def extract_data():
    print("Extracting data")
    return True


def validate_data():
    print("Validating data")
    return False


def transform_data():
    print("Transforming data")
    return True


def run_workflow():
    if not extract_data():
        print("Workflow stopped")
        return

    if not validate_data():
        print("Validation failed")
        print("Workflow stopped")
        return

    if not transform_data():
        print("Transformation failed")
        print("Workflow stopped")
        return

    print("Workflow completed")


run_workflow()
````

### 8. Stop the workflow if validation fails
This is an important orchestration concept.
- If validation fails, the transformation and loading tasks should not execute.

````python
def extract_data():
    print("Data extracted successfully")
    return True


def validate_data():
    print("Data validation failed")
    return False


def transform_data():
    print("Data transformed successfully")
    return True


def load_data():
    print("Data loaded successfully")
    return True


def run_pipeline():

    if not extract_data():
        print("Pipeline stopped")
        return

    if not validate_data():
        print("Pipeline stopped because validation failed")
        return

    if not transform_data():
        print("Pipeline stopped")
        return

    load_data()


run_pipeline()
````

## 🎤 Day 1: Introduction to Workflow Orchestration

Hello everyone 👋

Today I learned about **Workflow Orchestration**.

🔹 A workflow is a series of tasks that are completed in a specific order.

📝 For example, in a data pipeline, we can first extract the data, then clean it, transform it, and finally load it into a database.

📥 **Extract Data**  
↓  
🧹 **Clean Data**  
↓  
🔄 **Transform Data**  
↓  
📤 **Load Data**

⚙️ Workflow orchestration means managing these tasks and making sure they run in the correct order.

🤖 It helps Data Engineers automate pipelines instead of running every task manually.

🔗 It also manages dependencies between tasks. For example, the transformation task should not start until the data extraction task is completed.

🛡️ Orchestration can also help with failures. If a task fails, the workflow can stop, retry the task, or notify the Data Engineer.

💻 Today, I used Python to create simple tasks and execute them as a workflow.

🎯 My main learning today is that **workflow orchestration helps Data Engineers manage complex data pipelines in an automated and organized way**.
